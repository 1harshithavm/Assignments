```python
#assignment (24/03/2026)
#Assignment Name :Word Importance Explorer


from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Step 1: Create 5 documents
documents = [
    "Machine learning is very useful in data science",
    "Python is widely used in machine learning",
    "Data science uses machine learning techniques",
    "Artificial intelligence and machine learning are related",
    "Python programming is important for data science"
]

# Step 2: Apply TF-IDF
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(documents)

# Step 3: Convert to DataFrame
df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("TF-IDF Matrix:\n")
print(df)

# Step 4: Identify Top Keywords
print("\nTop Keywords in Each Document:\n")

for i in range(len(documents)):
    top_word = df.iloc[i].idxmax()
    print("Document", i+1, "Top Keyword:", top_word)
```

```python
#assignment 26/03/2026
#Assignment Name : Movie Review Analyzer


from textblob import TextBlob

# Step 1: Create movie reviews
reviews = [
    "I love this movie, it was amazing!",
    "This film was terrible and boring.",
    "The acting was fantastic and enjoyable.",
    "Worst movie I have ever seen.",
    "The movie was okay, not bad."
]

# Step 2: Analyze sentiment
for review in reviews:

    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity

    # Step 3: Classify sentiment
    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    # Step 4: Print results
    print("Review:", review)
    print("Sentiment:", sentiment)
    print("Polarity Score:", polarity)
    print()
```
